```python
def convert_distance():
    unit = input("Unit you are converting from (km, mile): ").strip().lower()
    value = float(input("Enter value to convert: "))

    if unit == "km":
        miles = value * 0.621
        print(f"{miles:.3f} miles")
    elif unit == "mile":
        kms = value / 0.621
        print(f"{kms:.3f} kms")
    else:
        print("Invalid unit! Please enter 'km' or 'mile'.")


convert_distance()
```
